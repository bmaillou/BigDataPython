### What is the inventory?

It is the total amount of lead in your core above background levels.  It is normalized to the area and has units of g/m$^2$.  No anthropenic lead added to sediment would equal 0.  Here is my AI schematic.  



```python
Image(filename='Invnetory_Schematic.jpeg',width=800)
```

How do I calculate?
-  For each layer of the core
-  You multiply the weight of the sediment by the concentration of lead.  In Steve's paper they subtract a background concentration of 26 ug/g from the concentrations at first which is the same as mg/Kg as this is the natural levels in soils.  This is stated in figure caption 3.  
-  Lead is in units of mg/Kg which is mg of lead per kg of soil
-  the soil weight is in grams so you convert to kg.  
-  this gives you the weight of the lead in each section in units mg.  
-  You sum of the weights
-  You convert to grams
-  you divide by the cross-sectional area of the core in units of m$^2$
-  the core radius is 1.27 cm


As math this looks like (watch the units!)

### $Inventory = \frac{1}{CoreArea}*\sum_{n=1}^{Sections}SedMass*(ConcentrationPb-26)$

### $Core Area= \frac{1.27^2*\pi}{100*100}=0.000507$

## Lead Concentrations versus Lead Isotopes

In order to understand the data you need to understand a little bit about lead isotopes.  We are using total lead and lead isotopes in two ways in this study and in this paper.  
- Lead Concentrations.  Total amount of Lead.  this controls the health effects.  
- Lead Isotopes.  These are the radionuclides of lead that can be used for dataing sediment.
  
Lead-210 is a natural radionuclide of lead and has a half life of 22 years.  Lead is deposited at basically a constant rate from the atmosphere.  But once sediment is buried it can't get anymore Lead-210.  So the lead-210 decays and after 22 years you will have half as much as when you started. So by measuring lead-210 you can get an idea of the time since depositions!  That is how Steve has dates on the core. The lead-210 used for dating is also called Excess Lead-210. 

But at the same time there is total Lead in the atmosphere.  This is all the stable and unstable isotopes of lead.  The amount of this total lead has changed in the atmosphere over time and we are interested in its temporal patterns and its controls.  So Lead-210 is helping with dating the sediment and total lead is telling us about atmospheric contamination.  

Also, within these cores there is deposition of Cs-137 from nuclear fallout.  Also, there are other elements and metals that may have a contamination source while many elements are just natural.  Can we figure out which are natural and which correlate with the different sources of contamination?

```python
from io import BytesIO
import time
import ipynbname
from ipylab import JupyterFrontEnd
from pypdf import PdfReader, PdfWriter
from reportlab.pdfgen import canvas

# 1. Force JupyterLab to save latest changes
app = JupyterFrontEnd()
app.commands.execute("docmanager:save")
time.sleep(1)

# 2. Get current notebook name
filename = ipynbname.name()

# 3. Convert notebook to PDF
!uv run jupyter nbconvert --to webpdf "{filename}.ipynb" --log-level=CRITICAL

# 4. Read PDF and isolate non-setup pages
reader = PdfReader(f"{filename}.pdf")
valid_pages = reader.pages[:-2]  # Strip final page
total_pages = len(valid_pages)


# Helper function to generate page number overlay
def create_footer_overlay(page_num, total, width, height):
  packet = BytesIO()
  c = canvas.Canvas(packet, pagesize=(width, height))
  c.setFont("Helvetica", 9)
  c.setFillColorRGB(0.3, 0.3, 0.3)  # Muted grey text

  # Stamp 0.5 inches (36pt) from bottom-right corner
  c.drawRightString(width - 36, 36, f"Page {page_num} of {total}")
  c.save()
  packet.seek(0)
  return PdfReader(packet).pages[0]


# 5. Stamp numbers and build final PDF
writer = PdfWriter()
for idx, page in enumerate(valid_pages, start=1):
  w = float(page.mediabox.width)
  h = float(page.mediabox.height)

  # Stamp page number onto the current page
  overlay = create_footer_overlay(idx, total_pages, w, h)
  page.merge_page(overlay)
  writer.add_page(page)

writer.write(f"{filename}.pdf")
print(
    f"Successfully generated {filename}.pdf ({total_pages} pages with page numbers)."
)
```
