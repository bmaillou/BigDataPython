```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
```

```python
fig,ax=plt.subplots()
ax.plot(np.linspace(0,2*np.pi),np.cos(np.linspace(0,2*np.pi)))
x=np.array([2.5,   2.5, 3.8, 3.8, 2.5, 4.0, 4.0, 2.5])
y=np.array([-0.5, 0.45, 0.45, 0, 0,   0,  -0.5, -0.5])
ax.plot(x,y)
ax.set_xlabel('X')  #added in
ax.set_ylabel('Y')  #added in
title='B inside Cosine of X made using\nlinspace, np.array([]) and ax.plot\nand a few other functions\n(It is an ugly B but that is ok)'
ax.set_title(title)
fig.savefig('CosineAndB.jpg',bbox_inches='tight')
```
